```python
# Nur Strommesswerte berücksichtigen (ohne timestamp)
strom = electricity.drop(columns=["timestamp"])

# Fehlende Messwerte pro Gebäude
missing = strom.isna().sum()

# Vollständigkeit pro Gebäude in Prozent
vollstaendigkeit = (1 - missing / len(strom)) * 100

# Fünf Gebäude mit der geringsten Vollständigkeit
print(vollstaendigkeit.sort_values().head(5))

# Durchschnittliche Vollständigkeit aller Gebäude
print(f"Durchschnittliche Vollständigkeit: {vollstaendigkeit.mean():.2f} %")

# Insgesamt fehlende Strommesswerte
print(f"Insgesamt fehlende Messwerte: {missing.sum()}")
```

```python
# Strommesswerte für 2016 auswählen
strom_2016 = electricity[
    electricity["timestamp"].dt.year == 2016
].drop(columns=["timestamp"])

vollstaendigkeit_2016 = strom_2016.notna().mean() * 100

strom_2017 = electricity[
    electricity["timestamp"].dt.year == 2017
].drop(columns=["timestamp"])

vollstaendigkeit_2017 = strom_2017.notna().mean() * 100

vergleich = pd.DataFrame({
    "Gesamt (%)": vollstaendigkeit,
    "2016 (%)": vollstaendigkeit_2016,
    "2017 (%)": vollstaendigkeit_2017
})

gebaeude = [
    "Fox_education_Maureen",
    "Fox_education_Vernon",
    "Fox_assembly_Jerrod"
]

display(vergleich.loc[gebaeude].round(2))
```

```python
# Qualitätskriterien für die Strommessreihen von Fox
gueltige_gebaeude = (
    (vollstaendigkeit >= 95) &
    (vollstaendigkeit_2016 >= 90) &
    (vollstaendigkeit_2017 >= 90)
)

# Anzahl der Gebäude, die alle Kriterien erfüllen
anzahl_gueltig = gueltige_gebaeude.sum()

print(f"Geeignete Gebäude: {anzahl_gueltig} von {len(gueltige_gebaeude)}")

# Gebäude, die mindestens ein Kriterium nicht erfüllen
print("\nNicht geeignete Gebäude:")
print(gueltige_gebaeude[~gueltige_gebaeude].index.tolist())
```

```python
# 1. BDG2-Wetterdaten laden
bdg = pd.read_csv("../data/weather.csv")

fox = bdg[bdg["site_id"] == "Fox"].copy()

# Lokale Zeit (Arizona) in UTC konvertieren
fox["timestamp_utc"] = (
    pd.to_datetime(fox["timestamp"])
    .dt.tz_localize("America/Phoenix")
    .dt.tz_convert("UTC")
)

# 2. Archivierte Open-Meteo-Daten laden
with open(raw_path, "r", encoding="utf-8") as file:
    weather_raw = json.load(file)

om = pd.DataFrame(weather_raw["hourly"])

# Open-Meteo-Zeitstempel sind bereits UTC
om["timestamp_utc"] = pd.to_datetime(om["time"], utc=True)

# 3. Beide Datenquellen über UTC zusammenführen
vergleich = pd.merge(
    fox[["timestamp_utc", "airTemperature"]],
    om[["timestamp_utc", "temperature_2m"]],
    on="timestamp_utc",
    how="inner"
)

# 4. Ungültige und fehlende Temperaturwerte ausschließen
vergleich["airTemperature"] = pd.to_numeric(
    vergleich["airTemperature"], errors="coerce"
)
vergleich["temperature_2m"] = pd.to_numeric(
    vergleich["temperature_2m"], errors="coerce"
)

vergleich = vergleich.dropna(
    subset=["airTemperature", "temperature_2m"]
)

# 5. Temperaturabweichung berechnen
vergleich["abweichung"] = (
    vergleich["airTemperature"] - vergleich["temperature_2m"]
).abs()

mae = vergleich["abweichung"].mean()

print("Verglichene Stunden:", len(vergleich))
print(f"MAE: {mae:.2f} °C")
print(f"Maximale Abweichung: {vergleich['abweichung'].max():.2f} °C")

display(vergleich.head(10))
```

```python

import duckdb
from pathlib import Path

# Der bereits verwendete absolute Datenbankpfad
parquet_path = Path(db_path).parent / "training_corpus.parquet"
parquet_sql_path = parquet_path.as_posix().replace("'", "''")

with duckdb.connect(str(db_path), read_only=True) as con:
    con.execute("SET TimeZone = 'UTC'")

    con.execute(f"""
        COPY (
            SELECT
                e.building_id,
                b.site_id,
                e.timestamp,
                e.consumption,
                b.primaryspaceusage,
                b.sqm,
                w.air_temperature,
                w.relative_humidity,
                w.shortwave_radiation
            FROM energy e
            JOIN building b
                ON e.building_id = b.building_id
            LEFT JOIN weather w
                ON b.site_id = w.site_id
                AND e.timestamp = w.timestamp
        )
        TO '{parquet_sql_path}'
        (FORMAT PARQUET, COMPRESSION ZSTD)
    """)

    result = con.execute("""
        SELECT
            COUNT(*) AS rows,
            COUNT(*) - COUNT(consumption) AS missing_consumption,
            COUNT(*) - COUNT(air_temperature) AS missing_weather
        FROM read_parquet(?)
    """, [str(parquet_path)]).df()

print("Parquet gespeichert:", parquet_path)
display(result)
```
